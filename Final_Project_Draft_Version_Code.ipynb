## 📉 STEP 12 — Regression Analysis
**Platform: scikit-learn**  
Which factors *predict* stress level? We use linear regression on Dataset 1 and logistic regression on Dataset 2.

```python
# ── Linear Regression — Dataset 1 ─────────────────────────────────────────
# Encode binary columns
df1_reg = df1.copy()
le = LabelEncoder()
for col in ['Gender', 'Depression', 'Anxiety', 'Panic Attack']:
    df1_reg[col + '_enc'] = le.fit_transform(df1_reg[col])

features = ['Sleep Quality', 'Study Load', 'Headache Frequency',
            'Academic Performance', 'Extracurricular Activities',
            'Depression_enc', 'Anxiety_enc']

X1 = df1_reg[features]
y1 = df1_reg['Stress Level']

model1 = LinearRegression()
model1.fit(X1, y1)
y_pred1 = model1.predict(X1)
r2_1 = r2_score(y1, y_pred1)

coef_df = pd.DataFrame({'Feature': features, 'Coefficient': model1.coef_})
coef_df = coef_df.sort_values('Coefficient', key=abs, ascending=False)

fig17 = px.bar(
    coef_df,
    x='Coefficient',
    y='Feature',
    orientation='h',
    color='Coefficient',
    color_continuous_scale='RdBu_r',
    title=f' Linear Regression Coefficients — Predicting Stress Level (R² = {r2_1:.3f})',
    labels={'Coefficient': 'Effect on Stress Level', 'Feature': 'Predictor Variable'}
)
fig17.add_vline(x=0, line_dash='dash', line_color='black')
fig17.update_layout(plot_bgcolor='white', title_font_size=14)
fig17.show()

print(f'\n📌 R² Score: {r2_1:.3f} — The model explains {r2_1*100:.1f}% of variance in stress levels.')
print('\n📋 Coefficients (top predictors):')
for _, row in coef_df.iterrows():
    direction = 'increases' if row['Coefficient'] > 0 else 'decreases'
    print(f'  {row["Feature"]:30s}: {row["Coefficient"]:+.3f}  → {direction} stress')
```

```python
# ── Logistic Regression — Dataset 2 (Binary Stress: 1=Low, 2=High) ────────
from sklearn.metrics import classification_report

features2 = ['Depression', 'Sleep Quality', 'Study Load', 'Academic Performance',
             'Anxiety', 'Future Career Concerns', 'Peer Pressure',
             'Social Support', 'Living Conditions']

X2 = df2[features2]
y2 = df2['Stress Level']  # 1 or 2

model2 = LogisticRegression(max_iter=1000)
model2.fit(X2, y2)
y_pred2 = model2.predict(X2)
accuracy = (y2 == y_pred2).mean()

coef_df2 = pd.DataFrame({'Feature': features2, 'Coefficient': model2.coef_[0]})
coef_df2 = coef_df2.sort_values('Coefficient', key=abs, ascending=False)

fig18 = px.bar(
    coef_df2,
    x='Coefficient',
    y='Feature',
    orientation='h',
    color='Coefficient',
    color_continuous_scale='RdBu_r',
    title=f' Logistic Regression — Predicting High vs Low Stress (Accuracy = {accuracy*100:.1f}%)',
    labels={'Coefficient': 'Effect on P(High Stress)', 'Feature': 'Predictor Variable'}
)
fig18.add_vline(x=0, line_dash='dash', line_color='black')
fig18.update_layout(plot_bgcolor='white', title_font_size=13)
fig18.show()

print(f'\n📌 Model Accuracy: {accuracy*100:.1f}%')
print('\n📋 Classification Report:')
print(classification_report(y2, y_pred2, target_names=['Low Stress', 'High Stress']))
```

DuckDB lets you run SQL directly on pandas DataFrames — great for quick aggregations and filtering without writing complex pandas code.

```python
import duckdb

# DuckDB can query pandas DataFrames directly using their variable names
con = duckdb.connect()

# ── Query 1: Average stress per year of study ─────────────────────────────
print(' Query 1: Average stress & sleep quality by Year of Study')
q1 = con.execute("""
    SELECT
        "Year Of Study"        AS year,
        COUNT(*)               AS n_students,
        ROUND(AVG("Stress Level"), 2)   AS avg_stress,
        ROUND(AVG("Sleep Quality"), 2)  AS avg_sleep
    FROM df1
    GROUP BY "Year Of Study"
    ORDER BY year
""").df()
print(q1.to_string(index=False))

# ── Query 2: High stress students with depression ─────────────────────────
print('\n Query 2: Students with HIGH stress (4–5) AND depression')
q2 = con.execute("""
    SELECT
        Gender,
        "Year Of Study",
        "Stress Level",
        "Sleep Quality",
        Depression
    FROM df1
    WHERE "Stress Level" >= 4
      AND Depression = 'Yes'
    ORDER BY "Stress Level" DESC
""").df()
print(f'Found {len(q2)} students with high stress AND depression:')
print(q2.to_string(index=False))

# ── Query 3: Dataset 2 — Avg scores by stress group ───────────────────────
print('\n Query 3: Dataset 2 — Key variable averages by stress group')
q3 = con.execute("""
    SELECT
        CASE WHEN "Stress Level" = 1 THEN 'Low Stress' ELSE 'High Stress' END AS stress_group,
        COUNT(*) AS n,
        ROUND(AVG(Depression), 2)    AS avg_depression_score,
        ROUND(AVG("Sleep Quality"), 2) AS avg_sleep,
        ROUND(AVG("Study Load"), 2)  AS avg_study_load,
        ROUND(AVG(Anxiety), 2)       AS avg_anxiety_score
    FROM df2
    GROUP BY stress_group
""").df()
print(q3.to_string(index=False))
```

```python
# ── Visualise DuckDB Query 3 results ─────────────────────────────────────
q3_melt = q3.melt(
    id_vars='stress_group',
    value_vars=['avg_depression_score', 'avg_sleep', 'avg_study_load', 'avg_anxiety_score'],
    var_name='Metric', value_name='Average Score'
)
label_map = {
    'avg_depression_score': 'Depression Score',
    'avg_sleep': 'Sleep Quality',
    'avg_study_load': 'Study Load',
    'avg_anxiety_score': 'Anxiety Score'
}
q3_melt['Metric'] = q3_melt['Metric'].map(label_map)

fig19 = px.bar(
    q3_melt,
    x='Metric',
    y='Average Score',
    color='stress_group',
    barmode='group',
    color_discrete_map={'Low Stress': '#42A5F5', 'High Stress': '#EF5350'},
    title=' DuckDB Result: Key Metrics — Low vs High Stress Students (Dataset 2)',
    labels={'stress_group': 'Stress Group'},
    text='Average Score'
)
fig19.update_traces(texttemplate='%{text:.1f}', textposition='outside')
fig19.update_layout(plot_bgcolor='white', title_font_size=14)
fig19.show()
```

---

### 📌 Finding 1 — Sleep is the strongest single predictor of stress
Across both datasets, lower sleep quality consistently correlates with higher stress.  
Students rating sleep quality at 1–2 (poor) showed average stress levels **above 3.5 out of 5**.

### 📌 Finding 2 — Depression and anxiety multiply stress risk
35% of students reported depression and 34% reported anxiety.  
Students with depression had measurably higher average stress scores than those without.

### 📌 Finding 3 — Study load adds stress, but not as strongly as sleep deprivation
The regression model shows study load has a positive coefficient (more load → more stress),  
but sleep quality's effect size is larger — suggesting sleep is the more actionable lever.

### 📌 Finding 4 — Year 1 and Year 3 students show highest vulnerability
First-year students face transition stress; Year 3 students face heavy academic demands.  
These are the groups where targeted support would have the greatest impact.

---

> **"Because poor sleep is the strongest predictor of student stress — more so than workload alone — we recommend that universities introduce structured sleep health programmes and reduce scheduling of early-morning lectures for students in Years 1 and 3."**
